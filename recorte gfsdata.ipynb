```python
import xarray as xr
from pathlib import Path

#abrir archico original
ds = xr.open_dataset(Path.home() / "Downloads" / "gfs_data.nc")

#quedamos solo con variables que usa la app
rec = ds[["t", "sphum", "o3mr", "zh"]]

#guaradr recorte en un archivo nuevo
#comprimir cada variabe ala guardar
compresion = {"t": {"zlib": True},
              "sphum": {"zlib": True}, 
              "o3mr": {"zlib": True},
              "zh": {"zlib": True}}
rec.to_netcdf("gfs_recorte.nc", encoding=compresion)
```
